```python
gav = pd.read_csv('CYP3a4_ki_scores.csv')
gav1 = pd.DataFrame(gav)
```

```python
gav1 = gav1.drop(gav1[gav1['pValue'] == 0.00].index)
```

```python

from scipy.stats import pearsonr, spearmanr

pearson_corr, pearson_p = pearsonr(gav1['pValue'], gav1['neg_log10_affinity_M'])
spearman_corr, spearman_p = spearmanr(gav1['pValue'], gav1['neg_log10_affinity_M'])

print(f"Корреляция Пирсона: r = {pearson_corr:.4f}, p-value = {pearson_p:.4e}")
print(f"Корреляция Спирмена: ρ = {spearman_corr:.4f}, p-value = {spearman_p:.4e}")


```

```python
import seaborn as sns 
import matplotlib.pyplot as plt
# Предполагается, что gav1 — ваш DataFrame с колонками pValue (экспериментальные) и neg_log10_affinity_M (предсказанные)
sns.set(style="white", font_scale=1.2)

# Строим jointplot с гистограммами по осям
g = sns.jointplot(
    x="neg_log10_affinity_M",
    y="pValue",
    data=gav1,
    kind="scatter",
    color="blue",
    marginal_kws=dict(bins=40, fill=True, color="blue", alpha=0.7)
)

# Добавим диагональ y=x для наглядности
import numpy as np
lims = [
    np.min([gav1["neg_log10_affinity_M"].min(), gav1["pValue"].min()]),
    np.max([gav1["neg_log10_affinity_M"].max(), gav1["pValue"].max()])
]
g.ax_joint.plot(lims, lims, '--', color='black', linewidth=1)

g.set_axis_labels("Predicted affinity", "Experimental pK")
plt.suptitle("Correlation between Experimental and Predicted affinity (CYP3a4)", y=1.02)

plt.tight_layout()
plt.savefig("correlation_plots_cyp3a4_ic50.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Данные для таблицы
data = {
    "Dataset": [
        "CYP3a4_ic50_scores",
        "CYP3a4_ki_scores",
        "CYP2d6_ic50_scores",
        "CYP2d6_ki_scores"
    ],
    "Pearson r": [0.1093, 0.2152, -0.0269, 0.2462],
    "Pearson p-value": [9.4485e-14, 3.8262e-02, 1.8469e-01, 9.4352e-07],
    "Spearman ρ": [0.0831, 0.2435, -0.0379, 0.2090],
    "Spearman p-value": [1.5485e-08, 1.8698e-02, 6.1565e-02, 3.4126e-05]
}

# Создаём DataFrame
df = pd.DataFrame(data)

# Форматируем числа для отображения в таблице (опционально)
df_display = df.copy()
df_display["Pearson r"] = df_display["Pearson r"].map("{:.4f}".format)
df_display["Pearson p-value"] = df_display["Pearson p-value"].map("{:.2e}".format)
df_display["Spearman ρ"] = df_display["Spearman ρ"].map("{:.4f}".format)
df_display["Spearman p-value"] = df_display["Spearman p-value"].map("{:.2e}".format)

# Создаём фигуру и ось
fig, ax = plt.subplots(figsize=(10, 2.5))  # ширина и высота под таблицу

ax.axis('off')  # Отключаем оси

# Создаём таблицу на основе DataFrame
table = ax.table(
    cellText=df_display.values,
    colLabels=df_display.columns,
    cellLoc='center',
    loc='center'
)

table.auto_set_font_size(False)
table.set_fontsize(12)
table.scale(1, 1.5)  # масштабируем таблицу для удобочитаемости

# Сохраняем таблицу в файл
plt.savefig("correlation_table.png", bbox_inches='tight', dpi=300)
plt.show()
```
